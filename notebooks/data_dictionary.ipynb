```python
import pandas as pd

begin_inv = pd.read_csv("../data/begin_inventory.csv")
end_inv = pd.read_csv("../data/end_inventory.csv")
purchase_prices = pd.read_csv("../data/purchase_prices.csv")
vendor_invoice = pd.read_csv("../data/vendor_invoice.csv")
purchases = pd.read_csv("../data/purchases.csv")
sales_chunk = pd.read_csv("../data/sales.csv", nrows=20)

```

```python
# Create a dictionary of your DataFrames

datasets = {
    "begin_inventory": begin_inv,
    "end_inventory": end_inv,
    "purchase_prices": purchase_prices,
    "vendor_invoice": vendor_invoice,
    "purchases": purchases,
    "sales_sample": sales_chunk  # only 20 rows
}

# Create a reusable function for profiling
def profile_df(name, df):
    print(f"\n=================== {name} ===================")
    print("\nHEAD:")
    display(df.head())

    print("\nINFO:")
    display(df.info())

    print("\nDESCRIBE:")
    try:
        display(df.describe(include="all"))
    except:
        print("Describe failed due to mixed datatypes.")

    print("\nNULL VALUES:")
    display(df.isna().sum())

    print("\nUNIQUE VALUES (first 10 columns):")
    for col in df.columns[:10]:
        print(f"  {col}: {df[col].nunique()} unique values")

# Loop through all datasets

for name, df in datasets.items():
    profile_df(name, df)

```

# Data Dictionary & Table Understanding (Concise)

This section summarizes what each table represents (grain), key fields, and how they relate to the final vendor performance analysis.

---

## 1. begin_inventory
**Grain:** One row = Store–Brand–Size inventory level at the **start of the year**.

**Key Fields:** InventoryId, Store, Brand, onHand, Price, startDate  
**Use:** Compute average inventory & turnover; understand capital at start.

---

## 2. end_inventory
**Grain:** One row = Store–Brand–Size inventory level at the **end of the year**.

**Key Fields:** InventoryId, Store, Brand, onHand, Price, endDate  
**Use:** Calculate ending inventory & unsold capital.

---

## 3. purchase_prices
**Grain:** One row = Standard vendor **price list entry** per Brand–Size–Vendor.

**Key Fields:** Brand, Price (retail), PurchasePrice (standard cost), VendorNumber  
**Use:** Compare standard prices vs actual purchase cost; identify discounts.

---

## 4. vendor_invoice
**Grain:** One row = One **invoice per Purchase Order (PO)**.

**Key Fields:** VendorNumber, PONumber, InvoiceDate, Dollars, Freight  
**Use:** Understand billed cost, freight allocation, payment timing.

---

## 5. purchases
**Grain:** One row = **Line-level purchase transaction** for Store–Brand–Vendor.

**Key Fields:** InventoryId, VendorNumber, PONumber, PurchasePrice, Quantity, Dollars  
**Use:** Compute actual unit cost, bulk purchasing behavior, vendor performance.

---

## 6. sales
**Grain:** One row = Sales event for Store–Brand on a given date.

**Key Fields:** InventoryId, SalesQuantity, SalesDollars, SalesPrice, SalesDate, VendorNo  
**Use:** Calculate revenue, gross profit, margin %, turnover.

---

# 🔗 Key Relationships

- **InventoryId** → links inventory ↔ purchases ↔ sales  
- **Brand, Size** → align product across multiple tables  
- **VendorNumber / VendorNo** → link purchases, sales, purchase_prices, vendor_invoice  
- **PONumber** → connects purchases ↔ vendor_invoice  
- **Dates** (PODate, InvoiceDate, SalesDate) → allow monthly aggregation

---
